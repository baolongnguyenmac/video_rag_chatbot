# Convert Audio to Text

```python
import whisper
from pysrt import SubRipFile, SubRipItem, SubRipTime
```

```python
model = whisper.load_model("base")
subtitle = model.transcribe('../../data/video/The Android Tab/The Android Tab.mp3')
```

```python
subtitle['segments']
```

```python
import json

with open('bla.json', 'w') as fo:
    json.dump([{
        'start': seg['start'],
        'end': seg['end'],
        'text': seg['text']
    } for seg in subtitle['segments']], fo)
```
